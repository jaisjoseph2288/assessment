```python
##question 4.2
orders = [
    ("alan", "keyboard"),
    ("alan", "Phone"),
    ("abin", "mouse"),
    ("abin", "headphones"),
]

grouped_orders = {
    customer.title(): [
        item for name, item in orders if name == customer
    ]
    for customer, _ in orders
}

print(grouped_orders)
```

```python
##question 4.1
class order:
    def __init__(self, name, product):
        self.name = name
        self.product = product
        self.index=0

    def __iter__(self):
        return self
    def __next__(self):
        if self.index>=len(self.product):
          order = self.orders[self.index]
           self.index += 1
           return order
```

```python
##question 2

sales = [
    {
        "order_id": 1,
        "customer": "Alan",
        "item": "Laptop",
        "quantity": 1,
        "price_per_item": 55000,
        "order_date": "2026-08-01"
    },
    {
        "order_id": 2,
        "customer": "abin",
        "item": "Mouse",
        "quantity": 2,
        "price_per_item": 450.00,
        "order_date": "2026-08-02"
    },
    {
        "order_id": 3,
        "customer": "Rahul",
        "item": "Keyboard",
        "quantity": 1,
        "price_per_item": 1200,
        "order_date": "2026-08-03"
    }
]

total_revenue=0
for i in sales:
    total_revenue += i["price_per_item"]*i["quantity"]
print("revenue:",total_revenue)

higher_value=[]
for i in sales:
    total_value = i["price_per_item"] * i["quantity"]
    if total_value > 2000:
        higher_value.append(i)
print(higher_value)
```
