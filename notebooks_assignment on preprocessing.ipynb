##  missing values in each column

```python
for i in['Sale Price',
    'No of Bathrooms',
    'Flat Area (in Sqft)',
    'Lot Area (in Sqft)','Area of the House from Basement (in Sqft)','Living Area after Renovation (in Sqft)','Latitude',
    'Longitude']:
    data[i]=data[i].fillna(data[i].median())
```

## out layer

```python
df=pd.read_csv("House_Pricing.csv")
```

```python
df
```

```python
for i in['Sale Price',
    'No of Bathrooms',
    'Flat Area (in Sqft)',
    'Lot Area (in Sqft)','Area of the House from Basement (in Sqft)','Living Area after Renovation (in Sqft)','Latitude',
    'Longitude']:
    df[i]=df[i].fillna(df[i].median())
```

```python
plt.boxplot(df['Sale Price'])
plt.show()
```

```python
q1=np.percentile(df['Sale Price'],25)
q2=np.percentile(df['Sale Price'],50)
q3=np.percentile(df['Sale Price'],75)
```

```python
print(q1,q2,q3)
```

```python
iqr=q3-q1
iqr
```

```python
low_lim=q1-1.5*iqr
upr_lim=q3+1.5*iqr
print(low_lim,upr_lim)
```

```python
outlier_index = []
for i in df.index:
    if df.loc[i, 'Sale Price'] < low_lim or df.loc[i, 'Sale Price'] > upr_lim:
        outlier_index.append(i)
```

```python
len(outlier_index)
```

```python
df = df.drop(outlier_index)
```

```python
df.shape
```
